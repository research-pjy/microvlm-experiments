# 00 Environment smoke test
Assert conda env, build mini+arch_a, one forward pass, print param count.


```python
from microvlm.utils.env_check import require_microvlm_env
require_microvlm_env()
print("conda env: microvlm")
```

```python
import torch
from microvlm.models.config import NanoVLMConfig, count_parameters
from microvlm.models.nanovlm import build_nanovlm

cfg = NanoVLMConfig(encoder_arch="arch_a")
model = build_nanovlm(cfg)
model.eval()
images = torch.randn(1, 3, 224, 224)
tokens = torch.randint(0, cfg.vocab_size, (1, 8))
logits, _ = model(images, tokens)
n = count_parameters(model)
print("logits", tuple(logits.shape))
print("params", n)
print("Table 1 mini ≈ 5_000_000")
assert 1_000_000 < n < 20_000_000
assert logits.shape[-1] == cfg.vocab_size
```

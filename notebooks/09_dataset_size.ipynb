# 09 Dataset size scaling
Calls `microvlm.experiments.exp09_dataset_size.run_exp09_dataset_size` only.


```python
from microvlm.utils.env_check import require_microvlm_env
from microvlm.utils.hydra_compose import compose_config
from microvlm.experiments.exp09_dataset_size import run_exp09_dataset_size

require_microvlm_env()
cfg = compose_config(config_name="experiment/exp09")
df = run_exp09_dataset_size(cfg)
df
```

```python
import matplotlib.pyplot as plt

numeric = df.select_dtypes(include="number")
if numeric.shape[1] >= 1:
    ax = numeric.iloc[:, : min(3, numeric.shape[1])].plot(kind="bar", figsize=(8, 4))
    ax.set_title("experiment results")
    plt.tight_layout()
    plt.show()
else:
    display(df)
```
